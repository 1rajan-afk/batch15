```python
#mean values prones to outliers/noise for eg: 1000,4000,3000,5000,NAN


#fill numeric columns with median

#fill categorical columns with mode

#optimal : ensure bedrooms is integer after filling

```

```python
df["Age"] = df["Age"].fillna(df["Age"].median(numeric_only=True))
```

```python
df["Embarked"]= df["Embarked"].fillna(df["Embarked"].mode().iloc[0])
```

```python
df["Cabin"]= df["Cabin"].fillna(df["Cabin"].mode().iloc[0])
```
